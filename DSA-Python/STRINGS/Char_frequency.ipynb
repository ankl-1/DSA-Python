```python
'''Problem statement : Given a string, count how many times each character appears.
  Requirements
- Solve it using Python.
- For now, don't use collections.Counter or other shortcuts.
- Treat uppercase and lowercase characters as different.'''

def char_freq(s):
  count={}
  for ch in s:
    if ch in count:
      count[ch]+=1
    else:
      count[ch]=1
  return count
print(char_freq('google.com'))
print(char_freq('BbAa'))
print(char_freq('e'))
print(char_freq(''))

'''time complexity : O(n)
   Space complexity : O(k) where k is the distinct elements
   Approach         : Using dictionary'''
```
