Problem statement: Given a string s, return the string in reverse order without using Python's built-in slicing ([::-1]) or a built-in reverse function.

```python
def reverse_string(a):
  j=len(a)-1
  b=""
  while j>=0:
    b+=a[j]
    j-=1
  return b
print(reverse_string("hello"))
print(reverse_string("python"))
print(reverse_string("a"))
print(reverse_string(""))
# Time Complexity: O(n²) in the worst case
# Auxiliary Space Complexity: O(n)
# Approach: Traverse the string backwards and build a new string.
```
