# Problem statement
Check whether a given string is a **palindrome.**

```python
def palindrome_str(a):
  j=len(a)-1
  b=""
  while j>=0:
    b+=a[j]
    j-=1
  if (a==b):
    return True
  else:
    return False
print(palindrome_str("madam"))
print(palindrome_str("hello"))
print(palindrome_str("abba"))
print(palindrome_str("a"))
print(palindrome_str(""))
print(palindrome_str("abca"))
#Time complexity:O(n^2)
#Auxiliary space:O(n)
#Approach:Reverse string and compare
```

```python
def is_palin(a):
  j=len(a)-1
  i=0
  while i<j:
    if(a[i]!=a[j]):
      return False
    else:
      i+=1
      j-=1
  return True
print(is_palin("madam"))
print(is_palin("hello"))
print(is_palin("abba"))
print(is_palin("a"))
print(is_palin(""))
print(is_palin("abca"))
#Time complexity:O(n)
#Auxiliary space:O(1)
#Approach:Two pointers
#This is better while comparing the time and space complexities
```
