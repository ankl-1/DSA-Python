```python
""" Given a sorted array a, remove the duplicate elements in place so that each unique element appears only once at the beginning of the array.
 Return the number of unique elements, k """


def duplicate(a):
  if not a:
    return 0
  i,j=1,1
  while i<len(a):
    if(a[i]==a[j-1]):
      i+=1
    else:
      a[j]=a[i]
      i+=1
      j+=1
  return j

a=[1, 1, 2, 2, 3]
k=duplicate(a)
print("Modified array:", a)
print("Unique elements:", a[:k])
print("Number of unique elements:", k)



  #Time complexity is : O(n)
  #Space complexity is : O(1)
  #Approach:Two pointers to remove duplicates in place

```
