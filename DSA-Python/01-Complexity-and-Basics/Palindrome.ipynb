```python
def palindrome(a):
  i=0
  j=len(a)-1
  while i<j:
    if(a[i]==a[j]):
      i+=1
      j-=1
    else:
      print("Not palindrome")
      return
  print("Is a Palindrome")
# Time Complexity: O(n)
# Auxiliary Space Complexity: O(1)
palindrome([1, 2, 3, 2, 1])
palindrome([1, 2, 3, 4, 1])


```
