```python
def sum_pair(a,target):
  i=0
  j=len(a)-1
  while i<j:
    if(a[i]+a[j]==target):
      return True
    elif(a[i]+a[j]<target):
      i+=1
    else:
      j-=1
  return False
print(sum_pair([1, 2, 4, 6, 8, 9], 14))
print(sum_pair([1, 2, 4, 6, 8, 9], 20))
print(sum_pair([], 10))
#time complexity: O(n)
#Auxiliary space: O(1)

```
