```python
                                              # Time Complexity: O(n)
def find_max(a):
  b=a[0]                                      # Space Complexity: O(1)
  for i in range(1,len(a)):
    if b<a[i]:
      b=a[i]
  return b
def find_min(a):
  b=a[0]
  for i in range(1,len(a)):
    if b>a[i]:
      b=a[i]
  return b
num=[2,4,8,1,90]
print(f"max={find_max(num)}")
print(f"min={find_min(num)}")
```
