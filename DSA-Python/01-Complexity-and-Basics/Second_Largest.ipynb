```python
def second_largest(a):
  l=a[0]
  s_l=None
  for i in range(1,len(a)):
    if (l<a[i]):
      s_l=l
      l=a[i]
    elif (s_l is None or s_l<a[i]):
      s_l=a[i]
  if s_l is not None:
    return s_l
  else:
    return None
# Time Complexity: O(n)
# Auxiliary Space Complexity: O(1)
print(second_largest([1,2,3,4,5,6,7,8,9]))
print(second_largest([2,2,1]))
print(second_largest([1]))
```
